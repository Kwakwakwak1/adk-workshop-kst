```python
# @title Vertex AI SDK Initialization
import vertexai

PROJECT_ID = "qwiklabs-gcp-04-79b727f6c308"
LOCATION = "us-central1"
STAGING_BUCKET = "gs://agent-staging-bucket-kcmo"

vertexai.init(
    project=PROJECT_ID,
    location=LOCATION,
    staging_bucket=STAGING_BUCKET,
)
```

```python
# @title Simple Google Search Agent Initialization
from google.adk.agents import Agent
from google.adk.tools import google_search
from vertexai.preview.reasoning_engines import AdkApp

search_agent = Agent(
    name="Google",
    model="gemini-2.5-flash",
    description="You can search Google...",
    instruction="You are a helpful Chatbot.",
    tools=[google_search]
)

app = AdkApp(agent=search_agent)
```

```python
# @title Local Agent Query Execution
for event in app.stream_query(
    user_id="local-test-user",
    message="Give me a keto friendly recipe that involves chicken thighs.",
):
    print(event)
```

```python
# @title Remote Agent Deployment to Agent Platform
from vertexai import agent_engines

remote_agent = agent_engines.create(
    app,
    requirements=["google-cloud-aiplatform[agent_engines,adk]"],
)
print("Deployment complete! Remote agent resource created.")
```

```python
# @title Remote Agent Query Execution
for event in remote_agent.stream_query(
    user_id="agent-engine-test-user",
    message="Give me a keto friendly recipe that involves chicken thighs.",
):
    print(event)
```
